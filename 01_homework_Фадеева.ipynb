"Иванов И. И." -> должно нормализоваться до "Иванов И.И." (Фамилия + Инициалы без пробелов)  
"И. И. Иванов" - "Иванов И.И."

Подсказка: напишите отдельные регулярки для инициалов и фамилий, затем напишите еще одну регулярку, которая включает возможные комбинации и порядки. Используйте именованные группы, чтобы преобразовать части в каноничный порядок.
В коде не должно быть никаких дополнительных нормализаций (дополнительный re.sub или .replace и т.п)

```python
NAMES = [
    "Иванов И. И.",
    "И. И. Иванов",
    "Иванов И.И.",
    "А.С. Пушкин",
    "Пушкин А.С.",
    "Д. М. Шеберле",
    "М. Кюри",
    "Ж.К. Ван Дамм",
]
```

```python
pattern = re.compile(r"""
    ^\s*
    (?:
        ([А-ЯЁ][а-яё]+(?:\s[А-ЯЁ][а-яё]+)?)\s+
        ([А-ЯЁ])\.\s*
        ([А-ЯЁ])?\.?
      |
        ([А-ЯЁ])\.\s*
        ([А-ЯЁ])?\.?\s+
        ([А-ЯЁ][а-яё]+(?:\s[А-ЯЁ][а-яё]+)?)
    )
    \s*$
""", re.VERBOSE)

for name in NAMES:
    m = pattern.match(name)
    surname = m.group(1) or m.group(6)
    init1 = m.group(2) or m.group(4)
    init2 = m.group(3) or m.group(5)
    print(f"{surname} {init1}." + (f"{init2}." if init2 else ""))
```

Например, можно случайно научить модель слишком быстро сдаваться. Но это критически важно развивать, поскольку в будущем пространство вариантов, где ИИ может сойти с ума, бесконечно. Особенно пугает информация о том, что со временем модели научатся понимать, что их тестируют в лабораторных условиях, и начнут имитировать безопасное поведение во время проверок, но вернутся к деструктивным действиям сразу после завершения аудита.

Инцидент с линейкой Claude показал, что модель способна выходить из-под контроля при первой же технической ошибке людей. Текущую проблему разработчики в значительной степени решили, но в глобальном – нет. В статье прямо сказано, что обучение чрезвычайно мощных моделей будущего быть надежно согласованными остается нерешенной технической задачей. Возможности искусственного интеллекта продолжают расти гораздо быстрее, чем наши способности гарантировать его безопасность. Если мы хотим, чтобы мощные модели будущего были надежно согласованы, недостаточно просто реагировать на уже произошедшие инциденты. Нужно развивать и общую систему безопасности, и конкретные защитные механизмы, и делать это быстрее, чем растут возможности самих моделей. Иначе следующая ошибка конфигурации может обернуться не просто утечкой данных, а чем-то гораздо более серьезным.
